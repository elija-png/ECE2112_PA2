```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
```

```python
x_mean = X.mean()
x_std = X.std()
```

```python
X_normalized = (X-x_mean)/x_std
```

```python
print("The Array is: ", X)
print("The mean is:", x_mean)
print("Standard Deviation is:", x_std)
print("Normalized Array: ", X_normalized)
```

```python
np.save("X_normalized.npy",X_normalized)
```

**CUBES DIVISIBLE BY 4 PROBLEM**

```python
C = np.arange(1,101)**3
print(C)
```

```python
C = C.reshape(10,10)
print(C)
```

```python
div_by_4 = C[C%4==0]
print(div_by_4)
```

```python
np.save('divide_by_4.npy', div_by_4)
```

**ABOVE MEAN SQUARES PROBLEM**

```python
S = np.arange(1,37)**2

print(S)
```

```python
S = S.reshape(6,6)
print(S)
```

```python
S_mean = S.mean()
print(S_mean)
```

```python
filter_mean = S > S_mean
print(filter_mean)
```

```python
above_mean = S[S > S_mean]
print (above_mean)
```

```python
np.save('above_mean.npy',above_mean)
```
